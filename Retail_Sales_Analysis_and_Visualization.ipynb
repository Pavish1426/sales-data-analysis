```python

# REAL-WORLD RETAIL DATA ANALYSIS PROJECT

# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# LOAD DATASET
df = pd.read_csv("/content/retail_sales_realworld_dataset.csv")

# DISPLAY DATA
df.head()

# DATASET INFORMATION
df.info()

# STATISTICAL SUMMARY
df.describe()

# CHECK MISSING VALUES
df.isnull().sum()

# DATA CLEANING

# Remove duplicates
df = df.drop_duplicates()

# Convert Date column
df['Date'] = pd.to_datetime(df['Date'])

# Create Month column
df['Month'] = df['Date'].dt.month

# DATA ANALYSIS

# Total Sales by Region
df.groupby('Region')['Sales'].sum()

# Total Profit by Product
df.groupby('Product')['Profit'].sum()

# Average Sales by Product
df.groupby('Product')['Sales'].mean()

# VISUALIZATIONS

# 1. Sales by Region
sns.barplot(
    x='Region',
    y='Sales',
    data=df
)

plt.title("Sales by Region")

plt.show()

# 2. Profit by Product
sns.barplot(
    x='Product',
    y='Profit',
    data=df
)

plt.title("Profit by Product")

plt.show()

# 3. Monthly Sales Trend
df.groupby('Month')['Sales'].sum().plot()

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.show()

# 4. Correlation Heatmap
sns.heatmap(
    df.corr(numeric_only=True),
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()

# 5. Sales vs Profit
sns.scatterplot(
    x='Sales',
    y='Profit',
    hue='Region',
    data=df
)

plt.title("Sales vs Profit")

plt.show()

# 6. Sales Distribution by Product
sns.boxplot(
    x='Product',
    y='Sales',
    data=df
)

plt.title("Sales Distribution by Product")

plt.show()
```
