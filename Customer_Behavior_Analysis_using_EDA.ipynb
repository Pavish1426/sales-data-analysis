```python

# EXPLORATORY DATA ANALYSIS (EDA)

# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("/content/eda_customer_dataset.csv")

# DISPLAY DATA
df.head()

# DATASET INFORMATION
df.info()

# STATISTICAL SUMMARY
df.describe()

# CHECK MISSING VALUES
df.isnull().sum()

# DATA ANALYSIS

# Average Monthly Income by Gender
df.groupby('Gender')['Monthly_Income'].mean()

# Average Spending Score by City
df.groupby('City')['Spending_Score'].mean()

# Correlation Between Numerical Columns
df.corr(numeric_only=True)

# VISUALIZATIONS

# 1. Age Distribution
sns.histplot(df['Age'], bins=5)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Count")

plt.show()

# 2. Income by Gender
sns.barplot(
    x='Gender',
    y='Monthly_Income',
    data=df
)

plt.title("Monthly Income by Gender")

plt.show()

# 3. Spending Score by City
sns.barplot(
    x='City',
    y='Spending_Score',
    data=df
)

plt.title("Spending Score by City")

plt.show()

# 4. Correlation Heatmap
sns.heatmap(
    df.corr(numeric_only=True),
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()

# 5. Income vs Spending Score
sns.scatterplot(
    x='Monthly_Income',
    y='Spending_Score',
    hue='Gender',
    data=df
)

plt.title("Income vs Spending Score")

plt.show()

# BOXPLOT FOR OUTLIERS
sns.boxplot(
    x='Gender',
    y='Monthly_Income',
    data=df
)

plt.title("Income Distribution")

plt.show()
```
